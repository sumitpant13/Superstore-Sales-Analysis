# Importing Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

# Importing and Reading Dataset

```python
df = pd.read_csv('Superstore.csv',encoding='unicode_escape')
```

```python
df.head(3)
```

# Shape of the dataset

```python
print('Dataset contains',df.shape[0],'rows and',df.shape[1],'columns.')
```

# Null value Check

```python
df.isna().sum()
```

# Duplicates check

```python
df.duplicated().sum()
```

# Disrepancy check

```python
for i in df.select_dtypes(include = 'object').columns:
    print('\n')
    print(df[i].value_counts())
    print('\n')
```

# DataSet Information

```python
df.info()
```

```python
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

# Feature Engineering — Derived Columns

```python
df['year'] = df['Order Date'].dt.year
df['month'] = df['Order Date'].dt.month
df['month_name'] = df['Order Date'].dt.strftime('%b')
df['year_month'] = df['Order Date'].dt.to_period('M').astype(str)
```

```python
df[['year','month', 'month_name', 'year_month']].head(5)
```

# Clean columns names

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

# Final Check

```python
print("Final shape:", df.shape)
```

# Saving Cleaned csv file

```python
df.to_csv('superstore_clean.csv', index=False)
```
